### Module 3 – Exception Handling
Exception handling is nothing but deals with errors, allowing to program to run.

Definiton - Exception handling prevents you program from crashing by giving meaning full message to users make your code runnable

Errors are two type -

compile error - like our mistakes - : or P instead p, like our errors or syntax error - before execution
run time error - 100% correct syntax - error after execution
Common Built-in Exceptions Exception Description

ZeroDivisionError Raised when dividing by zero 10 / 0

ValueError Raised when a function gets a valid type but an invalid value int("abc")

TypeError Raised when an operation is performed on incompatible data types "10" + 5

NameError Raised when a variable is not defined print(x)

IndexError Raised when a list index is out of range lst[10]

KeyError Raised when a dictionary key is not found d["age"]

AttributeError Raised when an object has no requested attribute "hello".append(5)

FileNotFoundError Raised when a file does not exist open("abc.txt")

ImportError Raised when a module cannot be imported import xyzmodule

ModuleNotFoundError Raised when a module is not found import mymodul

```python
# using try and except for Zerodivision error

try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    result = a / b
    print("Result:", result)

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
try:
    num = int(input("Enter a number: "))

except ValueError:
    print("Please enter an integer.")

else:
    if num % 2 == 0:
        print("Even Number")
    else:
        print("Odd Number")
```

```python
# type error

try:
    result = [1, 2, 3] * "2" #str
    print(result)

except TypeError:
    print("List can only be multiplied by an integer.")
```

```python
# Value error
try:
    num = int(input("Enter a number: "))

except ValueError:
    print("Invalid input.")

else:
    print("Square =", num ** 2)
```

```python
# name error

try:
    name = "Thamaraiselvan"
    print(nam)

except NameError:
    print("Variable name is incorrect.")
```

```python
#Index error 

try:
    fruits = ["Apple", "Banana", "Mango"]
    print(fruits[-5])

except IndexError:
    print("Invalid negative index.")
```

```python
# Key error

try:
    marks = {
        "Math": 90,
        "Science": 85,
        "English": 88
    }

    subject = input("Enter subject: ")
    print("Marks:", marks[subject])

except KeyError:
    print("Subject not found.")
```

```python
# Attribute error

try:
    num = 100
    print(num.upper())

except AttributeError:
    print("Integer object has no attribute 'upper'.")
```

```python
# FileNotFound error

try:
    file = open("students.txt", "r")
    print(file.read())
    file.close()

except FileNotFoundError:
    print("File not found.")
```

```python
#ImportError
try:
    from math import square

except ImportError:
    print("Function does not exist in the module.")
```

```python
#ModuleNotFoundError
try:
    import panda

except ModuleNotFoundError:
    print("Module does not exist.")
```

```python
#code using try,except,else,raise and finally

try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero.")

    result = a / b

except ValueError:
    print("Please enter valid numbers.")

except ZeroDivisionError as e:
    print(e)

else:
    print("Division successful.")
    print("Result:", result)

finally:
    print("Program completed.")
```

```python
#code using try,except,else,raise and finally

try:
    attendance = int(input("Enter attendance percentage: "))

    assert attendance >= 0 and attendance <= 100, "Attendance must be between 0 and 100."

    if attendance < 75:
        raise ValueError("Not eligible for exam.")

except AssertionError as e:
    print(e)

except ValueError as e:
    print(e)

else:
    print("Eligible for exam.")

finally:
    print("Process completed.")
```

```python

# Custom Exceptions

class PasswordError(Exception):
    pass

try:
    password = input("Enter password: ")

    if len(password) < 8:
        raise PasswordError("Password must contain at least 8 characters.")

except PasswordError as e:
    print(e)

else:
    print("Password accepted.")
```
