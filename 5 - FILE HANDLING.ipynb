### Reading Text Files

Python's built-in open() function opens a file. The "r" mode reads an existing file.

Example: Read an entire file

1) READING TEXT FILES
Reading a text file means retrieving the data stored in a file without changing it.

Python uses open() in read mode ('r').

Methods:

read()      - Reads entire file
readline()  - Reads one line
readlines() - Reads all lines as a list

```python
with open("Demo.txt", "r", encoding="utf-8") as file:
    content = file.read()
    print(content)
```

The file must exist, or Python raises FileNotFoundError.

### Writing Text Files

Python supports several file modes.

Mode  -   Purpose

r     -   Read an existing file

w     -   Write, replacing existing content

a     -   Append to the end of a file

x     -  Create a new file; fail if it exists

r+    -  Read and write an existing file

b     -   Binary mode, combined with another mode

## Modes:

w - Write (overwrite)
a - Append(add at the end)
x - Create new file

```python
# Write to a file

with open("Demo.txt", "w", encoding="utf-8") as file:
    file.write("Welcome to Python!\n")
    file.write("Learning file handling.")
```

```python
#Append without deleting existing content

with open("Demo.txt", "a") as file:
    file.write("\nThis is a new line.")
```

```python
#Write multiple lines
lines = ["Python\n", "Machine Learning\n", "Deep Learning\n"]

with open("topics.txt", "w", encoding="utf-8") as file:
    file.writelines(lines)
```

### CSV Files

CSV stands for Comma-Separated Values. It is widely used to store tabular datasets.

```python
import csv

students = [
    ["Name", "Age", "Course"],
    ["Arun", 22, "Python"],
    ["Priya", 23, "AI"],
    ["Thamarai", 21, "ML"]
]

with open("students.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(students)
    
```

```python
#Reading a CSV file

with open("students.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

```python
#Using DictReader

with open("students.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row["Name"], row["Course"])
```

### 4. JSON Files

JSON (JavaScript Object Notation) stores structured information using key-value pairs, arrays, strings, numbers, booleans, and null values.

It is commonly used in APIs, application configurations, and AI pipelines.

```python
import json

employee = {
    "name": "Arun",
    "age": 25,
    "skills": ["Python", "SQL", "Machine Learning"]
}

with open("employee.json", "w", encoding="utf-8") as file:
    json.dump(employee, file, indent=4)
```

```python
import json

file = open("Demo.json", "r")
data = json.load(file)
file.close()

print(data)
```

### PICKLE FILES
Pickle stores Python objects directly in binary format.

Terms:

Serialization: Saving objects
Deserialization: Loading objects

Pickle serializes Python objects into binary files and restores them later. It can store dictionaries, lists, and many other Python objects.

```python
import pickle

model_info = {
    "algorithm": "Random Forest",
    "accuracy": 0.92
}

with open("model_info.pkl", "wb") as file:
    pickle.dump(model_info, file)
```

```python
with open("model_info.pkl", "rb") as file:
    loaded_info = pickle.load(file)

print(loaded_info)
```

### Context Manager (with)

A context manager automatically closes a file when the block finishes, including when an exception occurs.

```python
# to write a file

with open("sample1.txt", "w") as file:
    file.write("Hello World")
    file.write("\nThis is Thamarai")

print("File created successfully.")
```

```python
# to read a file

with open("sample1.txt", "r") as file:
    data = file.read()

print(data)
```

### FILE VALIDATION
File validation checks whether

a file exists, 
is readable/writable,
is empty,
or has the correct type before using it.

```python
from pathlib import Path

def validate_csv(file_path, max_size_mb=10):
    path = Path(file_path)

    if not path.is_file():
        return False, "File does not exist"

    if path.suffix.lower() != ".csv":
        return False, "Only CSV files are allowed"

    if path.stat().st_size == 0:
        return False, "File is empty"

    if path.stat().st_size > max_size_mb * 1024 * 1024:
        return False, "File exceeds size limit"

    return True, "Basic validation successful"


valid, message = validate_csv("students.csv")

print(message)
```

Extension validation alone does not guarantee that a file contains valid CSV data. For stronger validation, inspect the contents and check the required columns.

```python
import csv

required_columns = {"Name", "Age", "Course"}

with open("students.csv", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    if reader.fieldnames and required_columns.issubset(reader.fieldnames):
        print("Required columns are present")
    else:
        print("Missing required columns")
```
