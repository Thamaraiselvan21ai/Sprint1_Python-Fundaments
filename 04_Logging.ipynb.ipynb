### Logging

Logging is the process of recording events, messages, warnings, and errors that occur while a program is running.

Python provides a built-in module called logging, so you don't need to install any additional packages.

For example, imagine an application that loads a machine learning model. You may want to record when the application starts, when the model loads successfully, and whether any errors occur.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Application started")
logging.info("Loading ML model")
logging.warning("Model is taking longer to load")
logging.error("Model loading failed")
```

### Why? 
Logging helps developers understand what is happening inside an application without constantly monitoring it.

Debugging: Identify where and why a program fails.

Error tracking: Record exceptions and failures.

Monitoring: Track application activity and performance.

Troubleshooting: Investigate problems using previously recorded events.

Auditing: Maintain records of relevant application operations.

### print() vs logging
print()
Used for simple output and quick checks.

logging
Used for structured message tracking in real applications.

Main difference
print() shows output directly. Logging gives control over:

importance of messages
where messages go
how messages look

```python
#Example using print()
print("Application started")
print("User logged in")
print("Something went wrong")
```

```python
#Example using logging
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format="%(levelname)s - %(message)s"
)

logging.info("Application started")
logging.info("User logged in")
logging.error("Something went wrong")
```

```python
import os
import logging

os.makedirs("logs", exist_ok=True)

logging.basicConfig(
    filename="logs/application.log",
    level=logging.INFO,
    format="%(levelname)s - %(message)s"
)

logging.info("Notebook execution started")

print("Log file created successfully!")
```

### Log Levels
DEBUG
Detailed information for debugging.

INFO
Normal program events.

WARNING
Something unexpected happened, but the program can continue.

ERROR
A serious problem happened in one operation.

CRITICAL
A very serious problem that may stop the program.

Formatter
A formatter decides how the log message should look.

Example: 2026-07-10 15:10:00 | INFO | Application started

A formatter makes logs:

readable
consistent
useful for debugging


```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format="%(levelname)s: %(message)s",
    force=True
)

logging.debug("Debugging information")
logging.info("Application started")
logging.warning("Low disk space")
logging.error("File not found")
logging.critical("System failure")
```

### Handlers
A handler decides where the log message goes.

### Console Handler
A console handler displays log messages in the terminal, command prompt, or notebook output.

Python provides logging.StreamHandler for this purpose. By default, it writes to standard error (stderr).

### File Handler
A file handler writes log messages to a file instead of displaying them only in the console.

This is useful for storing logs so developers can investigate errors after an application has finished running.

```python
logger = logging.getLogger("logging_example")
logger.setLevel(logging.DEBUG)

formatter = logging.Formatter("%(asctime)s | %(name)s | %(levelname)s | %(message)s")

console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)
console_handler.setFormatter(formatter)

file_handler = logging.FileHandler("logs/application.log")
file_handler.setLevel(logging.DEBUG)
file_handler.setFormatter(formatter)

logger.addHandler(console_handler)
logger.addHandler(file_handler)

logger.debug("Checking values")
logger.info("Task completed")
logger.warning("File is empty")
logger.error("File not found")
```

### Rotating Logs (Bonus)

When an application runs continuously, its log file can become very large. Rotating logs automatically create backup files when the current log file reaches a specified size.

Python provides RotatingFileHandler in the logging.handlers module.

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("rotating_example")
logger.setLevel(logging.INFO)
logger.propagate = False

handler = RotatingFileHandler(
    filename="rotating.log",
    maxBytes=1024,
    backupCount=3,
    encoding="utf-8"
)

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

handler.setFormatter(formatter)
logger.addHandler(handler)

for i in range(100):
    logger.info("Processing record %s", i)

handler.close()
logger.removeHandler(handler)
```

### Logging Best Practices

Follow these practices when developing Python applications.

Use named loggers: Prefer logging.getLogger(__name__) over using the root logger throughout your application.

Choose appropriate levels: Use DEBUG for diagnostic information, INFO for normal events, and ERROR for failed operations.

Include useful context: Record relevant IDs, filenames, operation names, and error details.

Log exceptions with tracebacks: Use logger.exception() inside exception handlers.

Never log sensitive information: Avoid recording passwords, API keys, authentication tokens, or unnecessary personal information.

Use lazy message formatting: Prefer logger.info("Processed %s records", count) instead of formatting the message beforehand.

Avoid duplicate handlers: Configure logging once, ideally at the application's entry point.

Rotate and retain logs appropriately: Prevent log files from growing indefinitely.
