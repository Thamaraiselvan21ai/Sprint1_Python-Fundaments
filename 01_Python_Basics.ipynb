## Control flow


### if/elif/else
In Python, the if-elif-else statement is used to make decisions in your code by executing different blocks of code based on specific conditions

* if checks the first condition.
* elif checks another condition if the first one is false.
* else runs when none of the conditions are true.
Only one matching block is executed.

```python
light=input("Enter the color:")
if light=="red":
    print("Stop")
elif light=="yellow":
    print("Ready")
else:
    print("Goo")
```

### for loop
Definition
A for loop is a control statement used to repeat a block of code for each item in a sequence or for a specific number of times.


```python
# A collection of 5 items
chocolates = ["Milk Chocolate", "Dark Chocolate", "Caramel", "Mint", "Hazelnut"]

# The For Loop
for candy in chocolates:
    print(f"Eating the {candy}!")
```

```python
# PrimeNumber

num = int(input("Enter a number: "))
count = 0
for i in range(1, num + 1):
    if num % i == 0:
        count += 1
if count == 2:
    print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```

### While loop
Definition
In Python, a while loop is used to repeatedly execute a block of code as long as a specified condition remains True. It is most useful when you don't know in advance how many times the loop will need to run.

```python
count = 1  # 1. Initialize a variable

while count <= 5:  # 2. Check the condition
    print(f"Count is {count}")
    count += 1  # 3. Update the variable
```

### Break
In Python, the break statement is used to immediately terminate a loop (either a for loop or a while loop). When Python encounters break, it stops executing the loop's code block and jumps directly to the first line of code following the loop.

```python
for i in range(1, 6):
    if i == 3:
        break
    print(i)
```

### Continue
The continue statement is used to skip the current iteration of a loop and move to the next iteration.


```python
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i)
```

### Pass
The pass statement is a placeholder. It tells Python: "Do nothing here for now."

It is useful when Python requires a statement, but you don't want to write the actual code yet.
