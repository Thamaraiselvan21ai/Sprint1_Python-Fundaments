### Module 2- OOPS - Object oriented programming 
OOPs is programming approach where we organize the code using class & objects.
The main OOP principle are:
1) Encapsulation - protect and organize the data
2) Abstraction - Hide unnecessary implementation details.
3) Inheritance  - reuse code from existing classes.
4) Polymorphism - Same interface, different behavior.


### 1. Class

A class is a blueprint/template for creating objects.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def result(self):
        if self.marks >= 35:
            print(self.name, "has Passed")
        else:
            print(self.name, "has Failed")

s1 = Student("Thamaraiselvan", 88)
s1.result()
```

```python
class Car:
    brand = "Toyota"
    color = "White"

car1 = Car()

print(car1.brand)
print(car1.color)
```

### Object
An Object is an instance of a class. It is the actual entity created using the blueprint defined by a class. When an object is created, memory is allocated for it, and it can store its own data while using the methods provided by the class.

```python
class Number:
    def __init__(self, num):
        self.num = num
    def check(self):
        if self.num % 2 == 0:
            print(self.num, "is Even")
        else:
            print(self.num, "is Odd")

n1 = Number(20)
n1.check()
```

**Self** represents the specific instance of a class that you are currently working with. It allows you to access and modify the attributes and methods belonging to that individual object, ensuring different instances of the same class do not overwrite each other's data.

### Constructor (init)
__init__() is automatically executed when an object is created.
It is commonly called the constructor in Python.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age
student1 = Student("Thamarai", 23)
student2 = Student("Arun", 22)

print(student1.name)
print(student2.name)     
```

### Instance Variables
Variables that belong to a particular object are called instance variables.

Instance Variables are variables that belong to an individual object of a class. They are usually initialized inside the __init__() constructor using the self keyword. Each object created from the same class has its own separate copy of these variables.

```python
class Student:
    def __init__(self, name, age):
        self.name = name   # Instance Variable
        self.age = age     # Instance Variable
student1 = Student("Thamarai", 23)
student2 = Student("Arun", 22)

print(student1.name)
print(student2.name)
```

### Class Variables
Class Variables are variables that belong to the class itself rather than to individual objects. They are declared directly inside the class but outside any methods. Since they are associated with the class, all objects created from that class share the same variable.
In short, A class variable is shared by all objects of a class.

```python
class Employee:
    company = "Infosys"      # Class Variable
    def __init__(self, name):
        self.name = name
    def display(self):
        print(self.name, "works at", Employee.company)

e1 = Employee("Raja")
e2 = Employee("Kiran")
e1.display()
e2.display()
```

### Instance Methods

Methods that operate on object data are called instance methods.
They use self.
**Instance Methods** are methods that operate on individual objects of a class. They are defined with the self parameter, which refers to the current object that calls the method. Through self, instance methods can access and modify the instance variables of that specific object.

```python
class Bank:
    def __init__(self, balance):
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
        print("Updated Balance =", self.balance)

b1 = Bank(5000)
b1.deposit(1000)
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def display(self):
        print(self.name)
        print(self.marks)
student = Student("Thamarai", 90)

student.display()
```

### Static Methods
Static Methods are methods that belong to a class but do not operate on instance variables or class variables. They are defined using the @staticmethod decorator and do not require the self or cls parameters.

```python
class Number:
    @staticmethod
    def check(num):
        if num % 2 == 0:
            print("Even Number")
        else:
            print("Odd Number")
Number.check(12)
```

### Class Methods
A class method works with the class itself.

Class Methods are methods that operate on the class itself rather than on individual objects. They are defined using the @classmethod decorator and receive the class as their first parameter, conventionally named cls.

```python
class Student:
    school = "ABC School"
    @classmethod
    def change_school(cl):
        cl.school = "XYZ School"

print(Student.school)
Student.change_school()
print(Student.school)
```

### Properties (@property)
 A Property in Python is a feature that allows a method to be accessed like an attribute using the @property decorator. It provides a controlled way to read, modify, or validate an object's data without exposing the internal implementation.

```python
class Employee:
    def __init__(self, salary):
        self.salary = salary
    @property
    def total_salary(self):
        return self.salary + 5000

e1 = Employee(40000)
print(e1.total_salary)
```

### Encapsulation
It is the process of combining data (attributes) and the methods (functions) that operate on that data into a single unit called a class. Encapsulation also controls direct access to an object's internal data by restricting how it can be viewed or modified.

```python

class BankAccount:

    def __init__(self, balance):
        self.__balance = balance      # Private Variable

    def deposit(self, amount):
        self.__balance += amount

    def show_balance(self):
        print("Balance =", self.__balance)


b1 = BankAccount(5000)

b1.deposit(1000)
b1.show_balance()
```

Note: protected (single underscore _),
private (double underscore __). 
* public      → name
* protected   → _name
* private     → __name

### Abstraction
Abstraction means hiding implementation details and exposing only what is necessary.

Python commonly uses the abc module.

```python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
        
#child class
class Dog(Animal):
    def sound(self):
        return "Bark"
dog = Dog()
print(dog.sound())
```

### Inheritance

Inheritance is one of the fundamental concepts of Object-Oriented Programming (OOP) that allows a new class to acquire the attributes and methods of an existing class. The existing class is called the Parent Class (Base Class), and the new class is called the Child Class (Derived Class).
Inheritance allows one class to reuse another class's properties and methods.

```python
#parent class
class Animal:
    def eat(self):
        print("Eating")

#Child class
class Dog(Animal):
    def bark(self):
        print("Barking")

dog = Dog()
dog.eat()
dog.bark()
```

### Multiple Inheritance
Multiple Inheritance is a type of inheritance in which a single child class inherits attributes and methods from two or more parent classes. This allows the child class to combine the functionalities of multiple classes into one, enabling greater code reuse and flexibility.

```python
class Father:
    def skills(self):
        print("Driving")

class Mother:
    def hobbies(self):
        print("Cooking")

class Child(Father, Mother):
    pass

child = Child()
child.skills()
child.hobbies()
```

### Multilevel Inheritance
Multilevel Inheritance is a type of inheritance in which a class inherits from another derived class, forming a chain of inheritance. In this hierarchy, one class acts as both a child of its parent class and a parent to another child class.
Inheritance happens across multiple levels.

```python
class Grandparent:
    def house(self):
        print("Grandparent house")


class Parent(Grandparent):
    def car(self):
        print("Parent car")


class Child(Parent):
    def bike(self):
        print("Child bike")
        
child = Child()
child.house()
child.car()
child.bike()
```

### Hierarchical Inheritance
Hierarchical Inheritance is a type of inheritance in which multiple child classes inherit from a single parent class. The parent class contains common attributes and methods that are shared among all child classes, while each child class can also define its own unique features and behaviors.

```python
class Animal:
    def eat(self):
        print("Animal Eats")


class Dog(Animal):
    def sound(self):
        print("Dog Barks")


class Cat(Animal):
 
    def sound(self):
        print("Cat Meows")


class Cow(Animal):

    def sound(self):
        print("Cow Moos")


class Lion(Animal):

    def sound(self):
        print("Lion Roars")


dog = Dog()
cat = Cat()
cow = Cow()
lion = Lion()

dog.eat()
dog.sound()

cat.eat()
cat.sound()

cow.eat()
cow.sound()

lion.eat()
lion.sound()
```

### Polymorphism
Polymorphism means:
One interface, different behavior.
Polymorphism is one of the core principles of Object-Oriented Programming (OOP). The word polymorphism means "many forms." It allows the same method, function, or interface to perform different actions depending on the object that uses it.

```python
class Dog:
    def sound(self):
        print("Dog Barks")


class Cat:
    def sound(self):
        print("Cat Meows")


class Cow:
    def sound(self):
        print("Cow Moos")


class Lion:
    def sound(self):
        print("Lion Roars")


dog = Dog()
cat = Cat()
cow = Cow()
lion = Lion()

dog.sound()
cat.sound()
cow.sound()
lion.sound()
```

### Method Overriding
Method Overriding is an Object-Oriented Programming concept in which a child class provides its own implementation of a method that is already defined in its parent class. The child class uses the same method name, the same parameters, and the same return type (where applicable), but changes the behavior according to its own requirements.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog Barks")

d = Dog()
d.sound()
```

Note: Python does not support traditional compile-time method overloading like Java/C++.
This will not work as expected:

```python
class Calculator:

    def add(self, a, b):
        return a + b

    def add(self, a, b, c):
        return a + b + c
calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
```

The second add() replaces the first.
**Python approach: Default Arguments**

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c
calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
```

```python
#Another approach: *args
class Calculator:
    def add(self, *numbers):
        return sum(numbers)

calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
print(calc.add(10, 20, 30, 40))
```

### Composition
Composition is an Object-Oriented Programming (OOP) concept in which one class contains an object of another class as one of its attributes. This represents a strong "has-a" relationship between the two classes. In composition, the contained object is created and managed by the owning class, and its lifecycle depends on the owner.

Composition represents a strong HAS-A relationship.
An object contains another object as a part of itself.

```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()

    def start_car(self):
        self.engine.start()
car = Car()
car.start_car()
```

Car HAS-A Engine

The engine is created inside the Car.

### Aggregation
Aggregation is also a HAS-A relationship, but the contained object can exist independently.

Aggregation is an Object-Oriented Programming (OOP) concept in which one class contains a reference to an object of another class, representing a weak "has-a" relationship. Unlike composition, the contained object is created independently and can continue to exist even if the parent object is destroyed. Aggregation allows multiple classes to share the same object without creating a strong dependency between them. This improves flexibility, promotes code reuse, and makes it easier to design loosely coupled systems.

```python
class Address:
    def city(self):
        print("City: Chennai")


class Person:
    def __init__(self, address):
        self.address = address

    def display(self):
        print("Person Details")
        self.address.city()


a = Address()
p = Person(a)
p.display()
```

| Composition                          | Aggregation                           |
| ------------------------------------ | ------------------------------------- |
| Strong HAS-A                         | Weak HAS-A                            |
| Objects are closely owned            | Objects can exist independently       |
| Component often created inside owner | Component usually supplied externally |


### Association

Association represents a relationship between independent objects.

Association is an OOPS concept that represents a relationship between two or more independent classes. In an association, the objects interact with each other to perform specific tasks, but neither object owns the other. Each object has its own lifecycle and can exist independently.

```python
class Student:

    def study(self):
        print("Student is studying")


class Teacher:

    def teach(self):
        print("Teacher is teaching")
student = Student()
teacher = Teacher()

teacher.teach()
student.study()
```

### super()
super() is a built-in Python function used to access the methods and constructors of a parent class from within a child class. It allows a child class to reuse the functionality of its parent class without directly referring to the parent's class name.

super() is used to access methods or attributes from a parent class.

```python
class Animal:
    def __init__(self, name):
        self.name = name


class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed
dog = Dog("Rocky", "Labrador")

print(dog.name)
print(dog.breed)
```

```python
class Animal:
    def show(self):
        print("This is an Animal")


class Dog(Animal):
    def show(self):
        super().show()
        print("This is a Dog")


d = Dog()
d.show()
```

### str()
The __str__() method is a special (magic or dunder) method in Python that defines the human-readable string representation of an object. It is automatically called whenever the print() function or the str() function is used on an object.

__str__() defines the human-readable representation of an object.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Student: {self.name}"

student = Student("Thamarai")

print(student)
```

### repr()
The __repr__() method is a special (magic or dunder) method in Python that defines the official string representation of an object. It is automatically called when the repr() function is used or when an object is displayed in the Python interpreter.

__repr__() provides a more detailed representation, primarily useful for developers/debugging.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Student(name='{self.name}', age={self.age})"
student = Student("Thamarai", 23)
print(repr(student))
```

### __len__()

__len__() allows an object to work with Python's len() function.

The __len__() method is a special (magic or dunder) method in Python that defines how the built-in len() function behaves for objects of a class. By implementing this method, developers can specify what should be considered the "length" of an object.

```python
class Team:

    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)
team = Team(["Arun", "Kumar", "Thamarai"])
print(len(team))
```

### Singleton Pattern
The Singleton Pattern is a software design pattern that ensures a class can have only one object (instance) throughout the lifetime of an application. Whenever an object of the Singleton class is requested, the same instance is returned instead of creating a new one.

The Singleton Pattern is useful when a single shared resource needs to be accessed from different parts of a program. Examples include database connections, application configuration, logging systems, cache managers, and thread pools.

Using the Singleton Pattern helps maintain consistency, reduces unnecessary memory usage, and prevents the creation of multiple objects that could lead to inconsistent application behavior.



```python
class Database:

    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.connected = False

        return cls._instance

    def connect(self):
        if not self.connected:
            print("Connecting to database...")
            self.connected = True
        else:
            print("Already connected")
db1 = Database()
db2 = Database()

db1.connect()
db2.connect()

print(db1 is db2)
```
